```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("1776250375-P2-Healthcare Analytics for Doctor Visits.csv")

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df.describe()
```

```python
gender_counts = df["gender"].value_counts()

print(gender_counts)
```

```python
gender_percentage = df["gender"].value_counts(normalize=True) * 100

print(gender_percentage.round(2))
```

Age Distribution

```python
print("Minimum age:", df["age"].min())
print("Maximum age:", df["age"].max())
print("Average age:", round(df["age"].mean(), 2))
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=10, kde=True)

plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel("Number of Patients")

plt.show()
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["0-18", "19-30", "31-45", "46-60", "60+"]
)

df["age_group"].value_counts().sort_index()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="age_group")

plt.title("Patients by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="age_group")

plt.title("Patients by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Patients")

plt.savefig("age_group_distribution.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
print("Minimum visits:", df["visits"].min())
print("Maximum visits:", df["visits"].max())
print("Average visits:", round(df["visits"].mean(), 2))
```

```python
visit_counts = df["visits"].value_counts().sort_index()

print(visit_counts)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="visits")

plt.title("Distribution of Doctor Visits")
plt.xlabel("Number of Doctor Visits")
plt.ylabel("Number of Patients")

plt.show()
```

```python
gender_visits = df.groupby("gender")["visits"].mean()

print(gender_visits.round(2))
```

```python
print("Minimum illness:", df["illness"].min())
print("Maximum illness:", df["illness"].max())
print("Average illness:", round(df["illness"].mean(), 2))
```

```python
illness_visits = df.groupby("illness")["visits"].mean()

print(illness_visits.round(2))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(data=df, x="illness", y="visits")

plt.title("Average Doctor Visits by Number of Illness Episodes")
plt.xlabel("Number of Illness Episodes")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
correlation = df["illness"].corr(df["visits"])

print("Correlation between illness and doctor visits:", round(correlation, 3))
```

```python
print(df["health"].value_counts().sort_index())
```

```python
health_visits = df.groupby("health")["visits"].mean()

print(health_visits.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="health", y="visits")

plt.title("Average Doctor Visits by Health Status")
plt.xlabel("Health Status")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
health_summary = df.groupby("health").agg(
    patients=("visits", "count"),
    average_visits=("visits", "mean")
)

print(health_summary.round(2))
```

```python
print("nchronic:")
print(df["nchronic"].value_counts())

print("\nlchronic:")
print(df["lchronic"].value_counts())
```

```python
nchronic_visits = df.groupby("nchronic")["visits"].mean()

print("Average visits by nchronic:")
print(nchronic_visits.round(2))
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="nchronic", y="visits")

plt.title("Average Doctor Visits by Chronic Condition")
plt.xlabel("Chronic Condition")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="lchronic", y="visits")

plt.title("Average Doctor Visits by Long-Term Chronic Condition")
plt.xlabel("Long-Term Chronic Condition")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
chronic_summary = df.groupby(["nchronic", "lchronic"])["visits"].agg(
    ["count", "mean"]
)

chronic_summary.columns = ["Patients", "Average Visits"]

print(chronic_summary.round(2))
```

```python
age_visit_summary = df.groupby("age_group", observed=True)["visits"].agg(
    ["count", "mean"]
)

age_visit_summary.columns = ["Patients", "Average Visits"]

print(age_visit_summary.round(2))
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="age_group",
    y="visits",
    errorbar=None
)

plt.title("Average Doctor Visits by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="age_group",
    y="visits",
    errorbar=None
)

plt.title("Average Doctor Visits by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Doctor Visits")

plt.savefig(
    "age_group_vs_doctor_visits.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
highest_age_group = age_visit_summary["Average Visits"].idxmax()
highest_average = age_visit_summary["Average Visits"].max()

print("Age group with highest average doctor visits:", highest_age_group)
print("Average doctor visits:", round(highest_average, 2))
```

```python
print("Minimum income:", df["income"].min())
print("Maximum income:", df["income"].max())
print("Average income:", round(df["income"].mean(), 2))
```

```python
df["income_group"] = pd.qcut(
    df["income"],
    q=4,
    labels=["Low", "Lower-Middle", "Upper-Middle", "High"]
)

df["income_group"].value_counts().sort_index()
```

```python
income_visit_summary = df.groupby(
    "income_group",
    observed=True
)["visits"].agg(["count", "mean"])

income_visit_summary.columns = ["Patients", "Average Visits"]

print(income_visit_summary.round(2))
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="income_group",
    y="visits",
    errorbar=None
)

plt.title("Average Doctor Visits by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Average Doctor Visits")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="income_group",
    y="visits",
    errorbar=None
)

plt.title("Average Doctor Visits by Income Group")
plt.xlabel("Income Group")
plt.ylabel("Average Doctor Visits")

plt.savefig(
    "income_vs_doctor_visits.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
highest_income_group = income_visit_summary["Average Visits"].idxmax()
highest_income_visits = income_visit_summary["Average Visits"].max()

print("Income group with highest average doctor visits:", highest_income_group)
print("Average doctor visits:", round(highest_income_visits, 2))
```

```python
correlation_matrix = df[
    ["visits", "age", "income", "illness", "reduced", "health"]
].corr()

print(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Healthcare Variables")

plt.show()
```

```python
visit_correlation = correlation_matrix["visits"].sort_values(
    ascending=False
)

print(visit_correlation.round(2))
```

```python
final_summary = df.groupby("gender").agg(
    Patients=("visits", "count"),
    Average_Visits=("visits", "mean"),
    Average_Illness=("illness", "mean"),
    Average_Age=("age", "mean")
)

print(final_summary.round(2))
```

```python
gender_avg = df.groupby("gender")["visits"].mean()

print("Average visits by gender:")
print(gender_avg.round(2))

print("\nGender with highest average visits:")
print(gender_avg.idxmax())
```

```python
age_avg = df.groupby("age_group", observed=True)["visits"].mean()

print("Average visits by age group:")
print(age_avg.round(2))

print("\nAge group with highest average visits:")
print(age_avg.idxmax())
```

```python
chronic_avg = df.groupby("nchronic")["visits"].mean()

print("Average visits by chronic condition:")
print(chronic_avg.round(2))

print("\nChronic condition group with highest average visits:")
print(chronic_avg.idxmax())
```

```python
illness_avg = df.groupby("illness")["visits"].mean()

print("Average visits by illness:")
print(illness_avg.round(2))

print("\nIllness level with highest average visits:")
print(illness_avg.idxmax())
```
